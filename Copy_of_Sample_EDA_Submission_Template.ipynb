## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Create a copy of the original dataset
df_clean = df.copy()

# 1. Remove completely duplicated rows
df_clean = df_clean.drop_duplicates()

# 2. Remove leading/trailing spaces from text columns
text_columns = df_clean.select_dtypes(include='object').columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

# 3. Convert empty strings into missing values
df_clean[text_columns] = df_clean[text_columns].replace('', np.nan)

# 4. Convert ID columns to appropriate numeric format
df_clean['person_id'] = pd.to_numeric(df_clean['person_id'], errors='coerce')

# 5. Keep the title ID as a string because IDs are identifiers,
# not numerical measurements
df_clean['id'] = df_clean['id'].astype('string')

# 6. Handle missing character values
# Character information may not be applicable to every role,
# so we replace missing values with "Unknown" rather than deleting records.
df_clean['character'] = df_clean['character'].fillna('Unknown')

# 7. Handle missing names
# Records without a person name cannot be meaningfully analyzed,
# so remove only those records.
df_clean = df_clean.dropna(subset=['name'])

# 8. Convert role to string and standardize text
df_clean['role'] = df_clean['role'].astype('string').str.strip()

# 9. Reset the index after cleaning
df_clean.reset_index(drop=True, inplace=True)

# Display the cleaned dataset
df_clean.head()
````

### What all manipulations have you done and insights you found?

Answer Here.
The following data-wrangling steps were performed to make the dataset analysis ready:

1. A copy of the original dataset was created so that the original data remained unchanged.

2. Duplicate records were removed to avoid repeated observations affecting the analysis.

3. Leading and trailing spaces were removed from the text columns to maintain consistency in categorical values.

4. Empty strings were converted into missing values so that they could be identified and handled properly.

5. The person_id column was converted into a numeric data type to maintain consistency for analysis.

6. The title id was treated as an identifier rather than a numerical measurement and was stored as a string.

7. Missing values in the character column were replaced with "Unknown". Character information may not be applicable to every type of role, so these records were retained instead of being unnecessarily deleted.

8. Records with missing person names were removed because a missing name makes the credit record difficult to use for person-level analysis.

9. The role column was standardized by removing unnecessary spaces.

10. The index was reset after the cleaning operations to maintain a continuous index.

Insights found from the data-wrangling process:

The cleaning process improved the consistency and quality of the dataset. Duplicate records and unnecessary spaces were handled, while important records with missing character information were retained. The analysis also showed that missing character values need to be interpreted carefully because the character field may not be relevant for every type of role.

These manipulations make the dataset more reliable and suitable for further Exploratory Data Analysis, statistical analysis, and visualization.

````python
# Average Credits per Contributor by Role

role_metrics = (
    df.groupby('role')
      .agg(
          total_credits=('person_id', 'size'),
          unique_contributors=('person_id', 'nunique')
      )
      .reset_index()
)

role_metrics['avg_credits_per_contributor'] = (
    role_metrics['total_credits'] /
    role_metrics['unique_contributors']
)

role_metrics = role_metrics.sort_values(
    'avg_credits_per_contributor',
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.plot(
    role_metrics['role'],
    role_metrics['avg_credits_per_contributor'],
    marker='o'
)

plt.title('Average Credits per Contributor by Role')
plt.xlabel('Role')
plt.ylabel('Average Credits per Contributor')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
````

The chart shows that the average number of credits per contributor varies across roles. Some roles have a higher average, indicating that contributors in those roles appear in more credit records on average, while other roles have a lower average.

This helps identify differences in contributor involvement across roles and provides a more detailed view than simply counting the total number of records.

Chart-18

````python
#  Percentage of Character Information Availability by Role

role_character_pct = pd.crosstab(
    df['role'],
    df['character'].isna().map({
        True: 'Missing',
        False: 'Available'
    }),
    normalize='index'
) * 100

role_character_pct.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Character Information Availability by Role (%)')
plt.xlabel('Role')
plt.ylabel('Percentage of Records')
plt.xticks(rotation=45)
plt.legend(title='Character Information')

plt.tight_layout()
plt.show()
````

1. Why did you pick the specific chart?

A 100% stacked bar chart was selected because it compares the proportion of available and missing character information within each role. Unlike a normal stacked bar chart, it allows a fair comparison between roles even when the total number of records in each role is different.


2. What is/are the insight(s) found from the chart?

The chart shows that the percentage of available and missing character information differs across roles. Some roles have a higher percentage of complete character information, while others have a greater proportion of missing values.

This highlights differences in metadata completeness across roles and helps identify roles that may require greater attention during data-quality improvement.
